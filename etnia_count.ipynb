```python
from models.gpkg import GeoPackage
from tools.exposure import spatial_join_pampa

gpkg = GeoPackage()
```

```python
ti = gpkg.read_layer('ti_all_revisada_v1')
bioma = gpkg.read_layer('biomas_rs_faixa_trans_final')
ti['pol_id'] = ti.index
```

```python
gdf = spatial_join_pampa(ti, bioma)
etnia_rename = {'Guaraní': 'Guarani', "Guaraní,Kaingang": 'Guarani - Kaingang', 'Guarani Mbya': 'Guarani','0': 'Sem dados', None: 'Sem dados'}
gdf.etnia_nome.replace(etnia_rename, inplace=True)
```

```python
table = gdf.groupby(['bioma', 'etnia_nome', 'fonte'])['etnia_nome'].count().reset_index(name='count')
```
